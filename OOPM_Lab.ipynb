#1. Write a Program to define a class and define member function inside the class.

```python
%%writefile object.cpp
#include <iostream>
using namespace std;

class Myclass{
    public:
    void display(){
        cout << "Hello World" << endl;
    }
};

int main(){
    Myclass obj;
    obj.display();
    return 0;
}
```

# 2. Write a program to define a class and define member function outside the class.

```python
%%writefile outobject.cpp
#include <iostream>
using namespace std;

class Rectangle{
    public:
    int length;
    int width;
    int area();
};

int Rectangle::area(){
    return length*width;
}

int main(){
    Rectangle rect;
    rect.length=5;
    rect.width=4;
    cout<< "Area of rectangle : "<< rect.area() << endl;
    return 0;
}
```

```python
!g++ outobject.cpp -o outobject
!./outobject
```

# 3. Write a program to implement type of object(Static, global, & dynamic).

```python
%%writefile stobject.cpp
#include <iostream>
using namespace std;

class Myclass{
    public:
    Myclass(){
        cout << "constructor called" << endl;
    }
    ~Myclass(){
        cout << "destructor called" << endl;
    }
    void display(){
        cout << "function called" << endl;
    }
};

Myclass global_obj;

int main(){
    cout <<"Main function starts" << endl;
    static Myclass static_obj;
    Myclass* dynamic_obj = new Myclass();
    global_obj.display();
    dynamic_obj->display();

    delete dynamic_obj;
    cout << "main function ends" << endl;

    return 0;
}
```

```python
!g++ stobject.cpp -o stobject
!./stobject
```

# 4. WAP to implement Array of Objects.

```python
%%writefile imobj.cpp
#include<iostream>
#include<string>
using namespace std;

class Student{
    public:
    string name;
    int age;
    void display(){
        cout<<"Name :"<<name<<", Age :"<<age<<endl;
    }
};

int main(){
    Student students[3];
    for(int i=0;i<3;i++){
        cout<<"Enter name for student "<<i+1<<":";
        cin>>students[i].name;
        cout<<"Enter age for student"<<i+1<<":";
        cin>>students[i].age;
    }
    cout<<"\nStudent Details:"<<endl;
    for (int i=0; i<3;i++){
        students[i].display();
    }
    return 0;
}
```

```python
!g++ imobj.cpp -o imobj
!./imobj
```
